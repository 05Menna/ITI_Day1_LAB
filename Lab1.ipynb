## 1. Experiment

```python
class Experiment:
    belongs_to = 'ITI AI Lab'
    def __init__(self,name,researcher,score):
        self.name = name
        self.researcher = researcher
        self.score = score

    def report(self):
        return f"{self.name} is done by {self.researcher} with score {self._score}"    

    @property
    def score(self):
        return self._score
    @score.setter
    def score(self,value):
        if 0 <= value <= 1:
            self._score=value
        else:
            raise ValueError("Score must be between 0 & 1")

    @classmethod
    def dict_to_object(cls,dictionary):
        return cls(dictionary["name"],dictionary["researcher"],dictionary["score"])


```

```python
experiment_data = { "name": "Sentiment Analysis",
                    "researcher": "Ahmed",
                    "score": 0.87
}               
exp = Experiment.dict_to_object(experiment_data)
print(exp.__dict__)
print(exp.report())
```

```python
exp1=Experiment("Sentiment","Menna",1)
print(exp1.belongs_to)
print(exp1.__dict__)
```

## 2. Different Experiment Types

### Classification Experiment

```python
class ClassificationExperiment(Experiment):
    def __init__(self, name, researcher, score, number_of_classes):
        super().__init__(name, researcher, score)
        self.number_of_classes= number_of_classes

    def report(self):
        return f"{self.name} is done by {self.researcher} with score {self._score} & number of classes are {self.number_of_classes}"    
```

```python
cexp=ClassificationExperiment("NLP","Mohamed",0.78,"7")
print(cexp.report())
```

### Retrieval Experiment

```python
class RetrievalExperiment(Experiment):
    def __init__(self, name, researcher, score, top_k):
        super().__init__(name, researcher, score)
        self.top_k=top_k

    def report(self):
        return f"{self.name} is done by {self.researcher} with score {self._score} & top {self.top_k} experiments"    
```

```python
rexp=RetrievalExperiment("Arabic RAG Search","Ahmed",0.88,4)
print(rexp.report())
```

## 3. Experiment Dashboard 

```python
class ExperimentDashboard:
    def __init__(self):
        self.experiments=[]
        
    def add_experiment(self,experiment):
        for exp in self.experiments:
            if exp.name == experiment.name:
                raise ValueError("This Experiment name is repeated")
        self.experiments.append(experiment)
        
    def show_reports(self):
        for exp in self.experiments:
            print(exp.report())
        print("")    

    def best_experiment(self):
        return max(self.experiments,key=lambda experiment: experiment.score)
          
    def structured_outcome(self):
        for exp in self.experiments:
            print(f" {exp.name} ({exp.score}),", end="")

        print(f" -> Best: {self.best_experiment().name}")  


```

```python
exp1 = Experiment("Sentiment Analysis","Menna", 0.87)
exp2 = ClassificationExperiment("Arabic RAG Search","Mohammed", 0.91,5)
exp3 = RetrievalExperiment("Image Classification","Ahmed", 0.83,4)
```

```python
expDash=ExperimentDashboard()

expDash.add_experiment(exp1)
expDash.add_experiment(exp2)
expDash.add_experiment(exp3)

expDash.show_reports()

print(f"Best: {expDash.best_experiment().name}")
```
